```python
i=5
for i in range(1,11):
    if(i==5):
        pass
        print(i)
else:
    print("completed")
```

```python
# using break
```

```python
i=10
for i in range(1,11):
    if(i==10):
        break
    print(i)
else:
    print("completed")
```

```python
# continue
```

```python
for i in range(1,11):
    if(i==2 or i==4 or i==10):
        continue
    print(i)
else:
    print("loop completed")
```
