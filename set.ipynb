```python
#add
fruits={"apple","mango","grapes","pine apple","papaya"}
fruits.add("orange")
fruits
```

```python
#remove-------> if element is not present ,it shows error
fruits.remove("orange")
fruits
```

```python
fruits.remove("kiwi")
fruits
```

```python
#discard
fruits.discard("apple")
fruits
```

```python
# if element is not present , it doesnt shows error
fruits.discard("grapes")
fruits
```

```python
# check membership-super fast
"mango" in fruits
```

```python
# length
len(fruits)
```

```python
# clear
fruits={"apple","mango","grapes","pine apple","papaya"}
fruits.clear()
fruits
```

```python
# why sets are powerful: math operations
a={1,2,3,4,5,}
b={4,5,6,7,8}
# union(a&b)-----> it combines 2 sets 
a.union(b)
```

```python
# intersection(a|b)----> we can pass aruments more than one using comma
# it shows wt we give the list along with same values in list
a={1,2,3,4,5,}
b={4,5,6,7,8}
c={9,10,11,12,13}
a.intersection(b),(c)
```

```python
# difference
#left-----> element not b , it shows
a.difference(b)
```

```python
#.right-----> element not in a , it shows
b.difference(a)
```

```python
#symmetric difference
a^b
```
