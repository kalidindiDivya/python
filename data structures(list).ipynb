```python
# list:
#List is a collection of homogenous and heterogenous data. 
    # list is a continuous memory location
    # it encloses with [] square brackets , each element seperated by ,(comma)
    # we can access elements based on the Index.
#SYNTAX----> variable(index)
# index supporting slicing 
# syntax---> variable(start:end:step)
```

```python
L1=[]
print(type(L1))
```

```python
L2=list()
print(type(L2))
```

```python
L3=[10,20,40,70,30,60]#---> Homogenous data
#l4=[1,"divya",3.14]----> heterogenous data
L3[1:5:1]
L3
```

```python
L3[5:1:-1]
L3
```

```python
# APPEND-----> TO ADD ELEMENT IN LAST 
L3.append(80)
L3
```

```python
# EXTEND-----> to  add more than 2 elements , single element is not consider
L3.extend([100,120,140])
L3
```

```python
L3.extend(150)
L3
```

```python
# DELETE-----> defaultly to remove last value in the collection
# we take as a variable as POP
L3.pop()
L3
```

```python
L3.pop(5)#-----> 5 IS INDEX VALUE
L3
```

```python
#REMOVE----> USED TO REMOVE ELEMENT THE COLLECTION OF DATA
L3.remove(100)
L3
```

```python
#DELETE
del L3[3]
L3
```

```python
#LENGTH
len(L3)
L3
```

```python
#SORTED-----> IT GIVES ASCENDING ORDER
sorted(L3)
L3
```

```python
#reverse sorted-----> it gives decending order
#SORT---> RETURNS NEW LIST
#SORTED---> RETURNS NEW LIST, ORIGINAL STAYS SAME
sorted(L3, reverse=True)
L3
```

```python
#ANY
any(L3)
L3
```

```python
#ALL
all(L3)
L3
```
